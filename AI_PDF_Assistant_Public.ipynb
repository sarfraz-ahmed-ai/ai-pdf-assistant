```python

import subprocess
import time

def check_ollama():
    return subprocess.run(
        ["ollama", "list"],
        capture_output=True,
        text=True
    )

result = check_ollama()

if result.returncode != 0:
    print("Starting Ollama server...")

    server = subprocess.Popen(
        ["ollama", "serve"],
        stdout=subprocess.DEVNULL,
        stderr=subprocess.DEVNULL
    )

    time.sleep(5)
    result = check_ollama()

if result.returncode == 0:
    print("OLLAMA READY")
    print(result.stdout)
else:
    print("OLLAMA NOT READY")
    print(result.stderr)
```

```python

chunk_size = 1000

chunks = [
    all_text[i:i + chunk_size]
    for i in range(0, len(all_text), chunk_size)
]

print("Total PDF characters:", len(all_text))
print("Total chunks:", len(chunks))
print("First chunk length:", len(chunks[0]))
print("Last chunk length:", len(chunks[-1]))
```

```python

keyword = "Konya"

matching_chunks = []

for index, chunk in enumerate(chunks):
    if keyword.lower() in chunk.lower():
        matching_chunks.append(index + 1)

print("Search keyword:", keyword)
print("Matching chunk numbers:", matching_chunks)
print("Total matches:", len(matching_chunks))
```

```python

keyword = "Konya"

for index, chunk in enumerate(chunks):
    count = chunk.lower().count(keyword.lower())
    print(f"Chunk {index + 1}: {count} occurrences")
```

```python

keyword = "Konya"

scores = [
    chunk.lower().count(keyword.lower())
    for chunk in chunks
]

best_index = scores.index(max(scores))
best_chunk = chunks[best_index]

print("Best chunk number:", best_index + 1)
print("Keyword occurrences:", scores[best_index])
print("Selected text length:", len(best_chunk))
```
